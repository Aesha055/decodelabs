```python
"""
DecodeLabs - Project 1: Rule-Based AI Chatbot
------------------------------------------------------
Goal: Build a deterministic, rule-based conversational agent.

Architecture (per design slides):
- INPUT LOOP      : Continuous 'while' cycle
- SANITIZATION    : Lowercase + whitespace normalization
- KNOWLEDGE BASE  : Dictionary mapping keys -> responses
- LOOKUP          : dict.get() handles match + fallback in one atomic op
- EXIT STRATEGY   : Clean break command

This is the "Logic Engine" - System 2 (deterministic, traceable, zero
hallucination risk). No ML, no probability - pure if/else and dictionary
lookups. Every output can be traced directly back to its input. This is
the "White Box" model: Input -> Logic -> Output, no mystery.
"""

# -------------------------------------------------
# KNOWLEDGE BASE
# -------------------------------------------------
# The "brain" of the bot. Each key is an exact-match intent;
# each value is the hard-coded response. Fully traceable - no
# hidden logic, no black box.
responses = {
    "hi": "Hello! How can I help you today?",
    "hello": "Hi there! What can I do for you?",
    "hey": "Hey! Good to see you.",
    "how are you": "I'm just a bunch of if-else statements, but I'm doing great! How about you?",
    "what is your name": "I'm DecodeBot, a rule-based AI assistant.",
    "who are you": "I'm DecodeBot — a deterministic, rule-based chatbot built at DecodeLabs.",
    "what can you do": "I can respond to greetings, answer simple questions, and chat using fixed rules. Try asking me something!",
    "help": "You can say hi, ask my name, ask how I am, or type 'bye' to exit.",
    "bye": "Goodbye! Have a great day!",
    "thanks": "You're welcome!",
    "thank you": "You're very welcome!",
}

# Default fallback response when no rule matches.
DEFAULT_RESPONSE = "I do not understand. Try typing 'help' to see what I can do."

# Words that trigger a clean exit from the loop.
EXIT_COMMANDS = {"bye", "exit", "quit"}


# -------------------------------------------------
# SANITIZATION
# -------------------------------------------------
def sanitize(text: str) -> str:
    """
    Normalize user input so lookups are consistent regardless of
    casing or stray whitespace.
    """
    return text.strip().lower()


# -------------------------------------------------
# CORE LOOKUP LOGIC
# -------------------------------------------------
def get_response(user_input: str) -> str:
    """
    The .get() method: handles lookup and fallback in a single
    atomic operation. No try/except, no nested if-chains needed.
    """
    cleaned = sanitize(user_input)
    return responses.get(cleaned, DEFAULT_RESPONSE)


# -------------------------------------------------
# MAIN INPUT LOOP
# -------------------------------------------------
def run_chatbot():
    print("=" * 60)
    print(" DecodeBot - Rule-Based AI Chatbot")
    print(" Type 'bye', 'exit', or 'quit' to end the conversation.")
    print("=" * 60)

    while True:  # Continuous input loop
        user_input = input("\nYou: ")
        cleaned = sanitize(user_input)

        # Empty input guard - re-prompt without crashing the loop
        if cleaned == "":
            print("DecodeBot: Please type something.")
            continue

        # Generate response BEFORE checking exit, so the bot can
        # still say goodbye instead of just dying silently.
        response = get_response(user_input)
        print(f"DecodeBot: {response}")

        # Clean exit strategy
        if cleaned in EXIT_COMMANDS:
            break

    print("\n[Session ended. DecodeBot has exited cleanly.]")


# -------------------------------------------------
# DEMO MODE (for non-interactive environments / quick testing)
# -------------------------------------------------
def run_demo():
    """
    Simulates a conversation without requiring manual input -
    useful for automated grading, notebooks, or quick sanity checks.
    """
    sample_inputs = [
        "Hi",
        "What is your name",
        "  HOW ARE YOU  ",
        "what can you do",
        "tell me a joke",   # unknown -> triggers fallback
        "thanks",
        "bye",
    ]

    print("=" * 60)
    print(" DecodeBot - Demo Mode (simulated conversation)")
    print("=" * 60)

    for raw in sample_inputs:
        print(f"\nYou: {raw}")
        print(f"DecodeBot: {get_response(raw)}")

    print("\n[Demo complete.]")


if __name__ == "__main__":
    import sys

    # Run demo mode automatically if no real terminal input is available
    # (e.g. when run in a sandbox), otherwise run the live chatbot.
    if sys.stdin.isatty():
        run_chatbot()
    else:
        run_demo()
```
